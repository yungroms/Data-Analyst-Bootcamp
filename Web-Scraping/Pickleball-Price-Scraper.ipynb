# Pickleball Paddle Price Tracker

An automated web scraper that tracks pricing for the Bread & Butter "Loco 16mm Pickleball Paddle" using `requests` and `BeautifulSoup`. 

The pipeline localizes data for the UK market (GBP currency and `DD/MM/YYYY` timestamps) and automatically appends records to a CSV file.

```python
# import libraries 

from bs4 import BeautifulSoup
import requests
import time
import datetime
import pandas as pd
```

```python
# Connect to Website and pull in data

URL = 'https://www.bnbpickleball.com/collections/loco/products/loco-16mm-pickleball-paddle-elongated'

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36...",
    "Accept-Language": "en-GB,en;q=0.9",  # Forces UK Welsh context
}
page = requests.get(URL, headers=headers)

soup1 = BeautifulSoup(page.content, "html.parser")
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")
```

```python
title = soup2.find('title')
print(title)

price = soup2.find('span', class_='current-price')
print(price)
```

```python
title = soup2.find('title').get_text().strip()
print(title)

price = soup2.find('span', class_='current-price').get_text().strip()
print(price)
```

```python
title = soup2.find('title').get_text().strip()[:42]
print(title)

price = soup2.find('span', class_='current-price').get_text().strip()
print(price)
```

```python
# Create a Timestamp for your output to track when data was collected

import datetime

today_us = datetime.date.today()
today = datetime.date.today().strftime('%d/%m/%Y') #UK formatted datetime

print(today_us)
print(today)
```

```python
# Create CSV and write headers and data into the file

import csv 

header = ['Title', 'Price', 'Date']
data = [title, price, today]


with open('PickleScraperDataset.csv', 'w', newline='', encoding='UTF8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)
```
