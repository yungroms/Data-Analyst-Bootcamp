# Web Scraping Pipeline: Largest US Companies by Revenue

## Objective:
Scrape real-time financial and corporate profile data from Wikipedia using `requests` and `BeautifulSoup`, clean the extracted text data, and structure it into a production-ready Pandas DataFrame for analysis and export.

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import os
```

## 1. Connecting and Fetching Target HTML
We send an HTTP GET request to Wikipedia using custom headers to ensure polite and compliant scraping.

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'

# Standard browser User-Agent header (generalized for portfolio publication)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36'
}

response = requests.get(url, headers=headers)
print(f"HTTP Status Code: {response.status_code}")

# Parse HTML tree
soup = BeautifulSoup(response.text, 'html.parser')
```

## 2. Parsing Table Headers and Rows
We target the primary data table (`wikitable`), extract the column headers (`<th>`), and iterate through the row cells (`<td>`) to capture company data.

```python
# Locate the primary target table on the page
table = soup.find_all('table')[0]

# Extract column titles
world_titles = table.find_all('th')
world_table_titles = [title.text.strip() for title in world_titles]

# Extract row-level data efficiently into a Python list
column_data = table.find_all('tr')[1:]
all_rows = []

for row in column_data:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
    if individual_row_data:
        all_rows.append(individual_row_data)

# Construct the Pandas DataFrame in a single step (avoids performance warnings)
df = pd.DataFrame(all_rows, columns=world_table_titles)

# Display a preview of the scraped dataset
display(df.head())
```

## 3. Exporting to CSV
We ensure a local `data/` directory exists and export the structured DataFrame for downstream analysis.

```python
os.makedirs('data', exist_ok=True)
output_path = 'data/companies.csv'

df.to_csv(output_path, index=False)
print(f"Dataset successfully compiled ({len(df)} records) and exported to '{output_path}'.")
```
