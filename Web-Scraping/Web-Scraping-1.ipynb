# Web Scraping Introduction: Extracting NHL Team Statistics

## Objective:
Demonstrate foundational web scraping techniques using `requests` and `BeautifulSoup` to extract structured tabular data from [ScrapeThisSite](https://www.scrapethissite.com/pages/forms/) and export it into a clean Pandas DataFrame.

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd

# Target URL for NHL team statistics practice
url = 'https://www.scrapethissite.com/pages/forms/'

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'
}

# Fetch page content
response = requests.get(url, headers=headers)
print(f"Status Code: {response.status_code}")

# Initialize BeautifulSoup parser
soup = BeautifulSoup(response.text, 'html.parser')
```

## 1. Extracting Table Headers
We locate all table header elements (`<th>`) to define the column names for our dataset.

```python
table = soup.find('table')
headers_tag = table.find_all('th')

column_titles = [title.text.strip() for title in headers_tag]
print("Extracted Columns:", column_titles)
```

## 2. Parsing Row Data & Building the DataFrame
We loop through all table rows (`<tr>`), extract individual data cells (`<td>`), clean whitespace, and compile the records into a Pandas DataFrame.

```python
# Initialize empty DataFrame with parsed column titles
df = pd.DataFrame(columns=column_titles)

# Extract all table rows (skipping the header row)
rows = table.find_all('tr')[1:]

for row in rows:
    cells = row.find_all('td')
    row_data = [cell.text.strip() for cell in cells]
    
    if row_data:
        df.loc[len(df)] = row_data

# Display the first few rows of the scraped dataset
display(df.head())

# Export dataset locally
df.to_csv('nhl_team_stats.csv', index=False)
print("Data successfully exported to nhl_team_stats.csv")
```
